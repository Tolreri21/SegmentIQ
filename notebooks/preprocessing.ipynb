```python
from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns

from segmentiq.clean import clean_df
from segmentiq.data import get_data
```

```python
root = Path.cwd().parents[0].resolve()
raw_path = root / Path("data/raw/Online Retail.xlsx")
cache_path = root / Path("data/cache/retail.parquet")
df = get_data(raw_path, cache_path)
```

```python
print(f"raw dataset: {df.shape}")
```

```python
cleaned_df = clean_df(df)
print(f"cleaned dataset: {cleaned_df.shape}")
```

## Cleaning result

We can see that 142220 rows were deleted. Most of them, 135080, are rows with missing CustomerID. The rest is 1916 rows with service stock codes, 5220 duplicates and 4 rows of the two biggest outliers with their cancellations.

```python
cleaned_df[["CustomerID", "Revenue"]].head()
```

```python
g = sns.boxplot(cleaned_df["Revenue"])
g.figure.suptitle("Box plot of the Revenue")
plt.show()
```

### Revenue

New column Revenue was successfully created and contains negative values. All of them are cancellations that I keep on purpose, because I will subtract them from the customer total instead of dropping them.

```python
print("missing values per column:")
print(cleaned_df.isna().sum())
print(f"duplicates: {cleaned_df.duplicated().sum()}")
```
